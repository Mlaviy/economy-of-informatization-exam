## 1. ИСХОДНЫЕ ДАННЫЕ И МАТЕМАТИЧЕСКАЯ МОДЕЛЬ

### Исходные параметры:
* Количество менеджеров ($N$): 15 чел.
* Средняя зарплата ($Z$): 120 000 руб./мес.
* Доля рутинных операций ($S_{rut}$): 30%
* Экономия времени на рутине ($E_{rut}$): 20%
* Рост продаж ($g$): 8% в год
* Текущий годовой оборот ($S_0$): 72 000 000 руб.
* Стоимость CRM на 1 чел./год ($C_{crm}$): 24 000 руб.
* Время на обучение ($T_{train}$): 1 месяц
* Горизонт оценки ($T$): 3 года
* Ставка дисконтирования ($r$): 12%
* Налог на прибыль ($t$): 20%

### Формулы:
1. Экономия времени: $\Delta T = S_{rut} \times E_{rut} = 30\% \times 20\% = 6\%$
2. Экономия ФОТ в год: $E_{fot} = N \times (Z \times 12) \times \Delta T$
3. Дополнительный доход в год: $\Delta S = S_0 \times g$
4. Разовые инвестиции (потери на обучение): $I_0 = N \times Z$
5. Затраты на CRM в год: $C_{year} = N \times C_{crm}$
6. Чистая прибыль в год: $NetProfit = (\Delta S + E_{fot} - C_{year}) \times (1 - t)$
7. $NPV = \sum_{k=1}^{T} \frac{NetProfit_k}{(1 + r)^k} - I_0$

```python
import pandas as pd
import numpy as np

# 1. ИСХОДНЫЕ ДАННЫЕ
n_managers = 15
salary_monthly = 120_000
routine_share = 0.30
routine_saving = 0.20

sales_growth = 0.08
annual_revenue = 72_000_000

crm_cost_per_user = 24_000
training_months = 1

horizon = 3
discount_rate = 0.12
tax_rate = 0.20

# 2. ПРЯМЫЕ ЭФФЕКТЫ И ЗАТРАТЫ
time_saving_ratio = routine_share * routine_saving  # 6%
annual_salary_savings = n_managers * (salary_monthly * 12) * time_saving_ratio
annual_extra_revenue = annual_revenue * sales_growth

annual_crm_cost = crm_cost_per_user * n_managers
initial_training_loss = n_managers * (salary_monthly * training_months)

# 3. ФИНАНСОВЫЙ РАСЧЕТ ПО ГОДАМ
years = list(range(1, horizon + 1))
extra_rev_list = [annual_extra_revenue] * horizon
salary_sav_list = [annual_salary_savings] * horizon
crm_cost_list = [annual_crm_cost] * horizon

ebit = [r + s - c for r, s, c in zip(extra_rev_list, salary_sav_list, crm_cost_list)]
taxes = [e * tax_rate for e in ebit]
net_profit = [e - t for e, t in zip(ebit, taxes)]

discount_factors = [1 / ((1 + discount_rate) ** y) for y in years]
discounted_cf = [np_val * df for np_val, df in zip(net_profit, discount_factors)]

# Показатели эффективности
npv = sum(discounted_cf) - initial_training_loss
total_net_profit = sum(net_profit) - initial_training_loss
total_investments = initial_training_loss + (annual_crm_cost * horizon)
roi = (total_net_profit / total_investments) * 100

# 4. ТАБЛИЦА И ВЫВОД РЕЗУЛЬТАТОВ
df = pd.DataFrame({
    'Год': years,
    'Доп. доход (руб.)': extra_rev_list,
    'Экономия ФОТ (руб.)': salary_sav_list,
    'Затраты CRM (руб.)': crm_cost_list,
    'Прибыль до налогов (EBIT)': ebit,
    'Налог на прибыль (20%)': taxes,
    'Чистая прибыль (руб.)': net_profit,
    'Дисконтированный поток (руб.)': discounted_cf
})

print(f"--- РЕЗУЛЬТАТЫ РАСЧЕТА ---")
print(f"Экономия на ЗП в год: {annual_salary_savings:,.2f} руб.")
print(f"Доп. доход от продаж в год: {annual_extra_revenue:,.2f} руб.")
print(f"Затраты на CRM в год: {annual_crm_cost:,.2f} руб.")
print(f"Потери на обучение (Инвестиции I0): {initial_training_loss:,.2f} руб.")
print(f"----------------------------------------")
print(f"NPV проекта: {npv:,.2f} руб.")
print(f"ROI проекта: {roi:.2f}%")
print(f"----------------------------------------")

display(df)
```
